### 6.3 Training Strategy (if applicable)

**Training Approach:** <span style="color:blue">Train from scratch (Supervised Learning using LightGBM)</span>

<span style="color:blue">I will use GridSearchCV to tune key LightGBM hyperparameters. The grid search will test different combinations of:

Minimum child samples (min_child_samples)

This approach will allow me to systematically search for strong-performing model settings while keeping computational cost manageable.</span>

### 7.2 Evaluation Methodology

**Train/Validation/Test Split:**

<span style="color:blue">I will split the dataset into 70% training, 15% validation, and 15% test sets. The split will be stratified by the target variable (IS_DELAYED) to ensure the proportion of delayed and on-time flights remains consistent across all datasets. The training set will be used to train the model and perform hyperparameter tuning. The validation set will be used to select the probability threshold and evaluate model performance during development. The test set will be held out and used only once at the end for final unbiased performance evaluation.</span>

**Cross-Validation Plan:**

<span style="color:blue">During hyperparameter tuning, I will use 3-fold cross-validation on the training data through GridSearchCV. This means the training set will be divided into three folds, and the model will train three times for each hyperparameter combination. If four hyperparameter combinations are tested, the model will be trained 12 total times (4 combinations × 3 folds). The performance across the folds will be averaged to select the best hyperparameters. Cross-validation will only be applied to the training set, not the validation or test sets.</span>

**Evaluation Frequency:**

<span style="color:blue">I will evaluate model performance at several stages of development. First, I will evaluate a baseline model on the validation set. During hyperparameter tuning, evaluation will occur automatically across the three cross-validation folds. After selecting the best hyperparameters, I will evaluate the tuned model on the validation set to determine the optimal probability threshold. Finally, I will evaluate the model once on the held-out test set to report final performance metrics. The test set will not be used for any tuning decisions to ensure an unbiased estimate of model performance.</span>
